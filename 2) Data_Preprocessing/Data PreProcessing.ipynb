```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

dataset['meal_name'] = le.fit_transform(dataset['meal_name'])
dataset['cuisine'] = le.fit_transform(dataset['cuisine'])
dataset['meal_type'] = le.fit_transform(dataset['meal_type'])
dataset['cooking_method'] = le.fit_transform(dataset['cooking_method'])
```
